```python

# ============================================
# PART II: DATA CLEANING AND FEATURE ENGINEERING
# ============================================

import pandas as pd
import numpy as np
from pathlib import Path

# 1. LOAD DATASET
if Path("facility_reservations.csv").exists():
    df = pd.read_csv("facility_reservations.csv")
elif Path("facility_reservations (1).xlsx").exists():
    df = pd.read_excel("facility_reservations (1).xlsx")
else:
    raise FileNotFoundError(
        "Upload facility_reservations.csv or facility_reservations (1).xlsx"
    )

# 2. PRESERVE ORIGINAL DATA
original_df = df.copy(deep=True)
df = df.copy()

print("Original dataset shape:", original_df.shape)

# 3. DATA-QUALITY SUMMARY BEFORE CLEANING
print("\n--- MISSING VALUES BEFORE CLEANING ---")
print(df.isnull().sum())

print("\n--- DUPLICATE ROWS BEFORE CLEANING ---")
print("Duplicate rows:", df.duplicated().sum())

print("\n--- DATA TYPES ---")
print(df.dtypes)

# 4. REMOVE EXACT DUPLICATE TRANSACTIONS
duplicates_removed = int(df.duplicated().sum())
df = df.drop_duplicates().copy()

# 5. STANDARDIZE TEXT CATEGORIES
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Standardize known categorical fields
for col in ["Facility_Type", "User_Category", "Usage_Status"]:
    if col in df.columns:
        df[col] = df[col].astype("string").str.strip().str.title()

# Normalize common status variants
if "Usage_Status" in df.columns:
    status_map = {
        "No Show": "No-Show",
        "Noshow": "No-Show",
        "No-Show": "No-Show",
        "Used": "Used",
        "Cancelled": "Cancelled",
        "Canceled": "Cancelled"
    }
    df["Usage_Status"] = df["Usage_Status"].replace(status_map)

# 6. CONVERT RELEVANT NUMERICAL COLUMNS
numeric_cols = [
    "Participants",
    "Actual_Attendance",
    "Lead_Time_Days",
    "Cancellation_Days_Before"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Correct invalid negative counts by marking them missing
for col in ["Participants", "Actual_Attendance"]:
    if col in df.columns:
        df.loc[df[col] < 0, col] = np.nan

# Participants must be positive for a valid attendance ratio
if "Participants" in df.columns:
    df.loc[df["Participants"] == 0, "Participants"] = np.nan

# Negative lead times are invalid
if "Lead_Time_Days" in df.columns:
    df.loc[df["Lead_Time_Days"] < 0, "Lead_Time_Days"] = np.nan

if "Cancellation_Days_Before" in df.columns:
    df.loc[df["Cancellation_Days_Before"] < 0,
           "Cancellation_Days_Before"] = np.nan

# Attendance cannot exceed expected participants when both are known
if {"Actual_Attendance", "Participants"}.issubset(df.columns):
    invalid_attendance = (
        df["Actual_Attendance"].notna()
        & df["Participants"].notna()
        & (df["Actual_Attendance"] > df["Participants"])
    )
    df.loc[invalid_attendance, "Actual_Attendance"] = np.nan

# 7. FEATURE ENGINEERING: LEAD TIME CATEGORY
if "Lead_Time_Days" in df.columns:
    conditions = [
        df["Lead_Time_Days"].between(0, 2),
        df["Lead_Time_Days"].between(3, 7),
        df["Lead_Time_Days"].between(8, 14),
        df["Lead_Time_Days"] > 14
    ]

    categories = [
        "Last Minute",
        "Short Lead",
        "Medium Lead",
        "Early Booking"
    ]

    df["Lead_Time_Category"] = np.select(
        conditions,
        categories,
        default="Unknown"
    )

    # Keep missing/invalid lead times missing rather than inventing values
    df.loc[df["Lead_Time_Days"].isna(),
           "Lead_Time_Category"] = pd.NA

# 8. FEATURE ENGINEERING: RESERVATION OUTCOME
if "Usage_Status" in df.columns:
    status = df["Usage_Status"].astype("string").str.strip().str.lower()

    df["Reservation_Outcome"] = pd.Series(
        pd.NA, index=df.index, dtype="string"
    )

    df.loc[status == "used", "Reservation_Outcome"] = "Completed"

    df.loc[
        status.isin(["cancelled", "no-show", "no show", "noshow"]),
        "Reservation_Outcome"
    ] = "Not Completed"

# 9. FEATURE ENGINEERING: ATTENDANCE RATIO
if {"Actual_Attendance", "Participants"}.issubset(df.columns):
    valid_denominator = df["Participants"].notna() & (
        df["Participants"] > 0
    )

    df["Attendance_Ratio"] = np.nan

    df.loc[valid_denominator, "Attendance_Ratio"] = (
        df.loc[valid_denominator, "Actual_Attendance"]
        / df.loc[valid_denominator, "Participants"]
    )

# 10. DATA-QUALITY SUMMARY AFTER CLEANING
print("\n--- CLEANING SUMMARY ---")
print("Original rows:", len(original_df))
print("Exact duplicate rows removed:", duplicates_removed)
print("Rows after duplicate removal:", len(df))
print("Columns after feature engineering:", len(df.columns))

print("\n--- MISSING VALUES AFTER CLEANING ---")
print(df.isnull().sum())

print("\n--- REMAINING DUPLICATE ROWS ---")
print(df.duplicated().sum())

# 11. SAVE CLEANED DATASET
df.to_csv("facility_reservations_cleaned.csv", index=False)

print("\nCleaned dataset saved as facility_reservations_cleaned.csv")

# 12. DISPLAY CLEANED DATA
display(df.head(10))
```

```python

# ============================================
# PART III: EXPLORATORY DATA ANALYSIS
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# 1. LOAD THE CLEANED DATASET
if Path("facility_reservations_cleaned.csv").exists():
    df = pd.read_csv("facility_reservations_cleaned.csv")
elif Path("facility_reservations.csv").exists():
    df = pd.read_csv("facility_reservations.csv")
elif Path("facility_reservations (1).xlsx").exists():
    df = pd.read_excel("facility_reservations (1).xlsx")
else:
    raise FileNotFoundError("Please upload the reservation dataset.")

print("Dataset shape:", df.shape)
print("Available columns:", df.columns.tolist())

# 2. CHECK REQUIRED COLUMNS
required = ["Facility_Type", "User_Category", "Usage_Status"]

missing = [col for col in required if col not in df.columns]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}. "
        "Check your dataset's actual column names."
    )

# Standardize text used in this analysis
for col in required:
    df[col] = df[col].astype("string").str.strip()

df["Usage_Status"] = df["Usage_Status"].str.lower().replace({
    "canceled": "cancelled",
    "no show": "no-show",
    "noshow": "no-show"
})

# ============================================
# QUESTION 1: WHICH FACILITY TYPE HAS THE
# HIGHEST RESERVATION COUNT?
# ============================================

facility_counts = (
    df["Facility_Type"]
    .dropna()
    .value_counts()
    .sort_values(ascending=False)
)

print("\n1. RESERVATION COUNT BY FACILITY TYPE")
print(facility_counts.to_string())

if not facility_counts.empty:
    top_facility = facility_counts.index[0]
    top_count = int(facility_counts.iloc[0])

    print(
        f"\nFinding: {top_facility} has the highest reservation "
        f"count, with {top_count} reservations."
    )

    plt.figure(figsize=(9, 5))
    facility_counts.plot(kind="bar")
    plt.title("Reservation Count by Facility Type")
    plt.xlabel("Facility Type")
    plt.ylabel("Number of Reservations")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("No valid facility-type data available.")

# ============================================
# QUESTION 2: WHICH USER CATEGORY HAS THE
# HIGHEST CANCELLATION/NO-SHOW PERCENTAGE?
# ============================================

valid_statuses = ["used", "cancelled", "no-show"]

status_df = df[
    df["Usage_Status"].isin(valid_statuses)
    & df["User_Category"].notna()
].copy()

status_df["Is_Cancelled_No_Show"] = (
    status_df["Usage_Status"].isin(["cancelled", "no-show"])
)

# Denominator = reservations with a recognized outcome
# for each user category
user_summary = status_df.groupby("User_Category").agg(
    Total_Reservations=("Is_Cancelled_No_Show", "size"),
    Cancelled_No_Show=("Is_Cancelled_No_Show", "sum")
)

user_summary["Cancellation_No_Show_Percent"] = (
    user_summary["Cancelled_No_Show"]
    / user_summary["Total_Reservations"]
    * 100
)

user_summary = user_summary.sort_values(
    "Cancellation_No_Show_Percent", ascending=False
)

print("\n2. CANCELLATION/NO-SHOW RATE BY USER CATEGORY")
print(user_summary.round(2).to_string())

if not user_summary.empty:
    highest_user = user_summary.index[0]
    highest_rate = user_summary.iloc[0][
        "Cancellation_No_Show_Percent"
    ]
    highest_n = int(user_summary.iloc[0]["Total_Reservations"])
    highest_events = int(user_summary.iloc[0]["Cancelled_No_Show"])

    print(
        f"\nFinding: {highest_user} has the highest cancellation/"
        f"no-show percentage at {highest_rate:.2f}% "
        f"({highest_events} cancelled/no-show reservations out of "
        f" {highest_n} reservations with recognized outcomes)."
    )

    plt.figure(figsize=(9, 5))
    user_summary["Cancellation_No_Show_Percent"].plot(
        kind="bar"
    )
    plt.title("Cancellation and No-Show Percentage by User Category")
    plt.xlabel("User Category")
    plt.ylabel("Cancelled/No-Show Reservations (%)")
    plt.xticks(rotation=45, ha="right")
    plt.ylim(0, 100)
    plt.tight_layout()
    plt.show()
else:
    print("No valid user-category outcomes available.")

# ============================================
# QUESTION 3: HOW DOES LEAD-TIME CATEGORY
# RELATE TO RESERVATION OUTCOME?
# ============================================

lead_df = df.copy()

# Support both possible column names for compatibility
lead_col = None
if "Lead_Days" in lead_df.columns:
    lead_col = "Lead_Days"
elif "Lead_Time_Days" in lead_df.columns:
    lead_col = "Lead_Time_Days"

if lead_col is not None:
    lead_df[lead_col] = pd.to_numeric(
        lead_df[lead_col], errors="coerce"
    )

    conditions = [
        lead_df[lead_col].between(0, 2),
        lead_df[lead_col].between(3, 7),
        lead_df[lead_col].between(8, 14),
        lead_df[lead_col] > 14
    ]

    lead_df["Lead_Time_Category"] = np.select(
        conditions,
        ["Last Minute", "Short Lead", "Medium Lead", "Early Booking"],
        default="Unknown"
    )

    lead_df.loc[
        lead_df[lead_col].isna(),
        "Lead_Time_Category"
    ] = pd.NA

if "Lead_Time_Category" not in lead_df.columns:
    raise ValueError(
        "Lead_Time_Category or a valid Lead_Days/Lead_Time_Days column is required."
    )

lead_df["Lead_Time_Category"] = (
    lead_df["Lead_Time_Category"].astype("string").str.strip()
)

lead_df["Reservation_Outcome"] = pd.Series(
    pd.NA, index=lead_df.index, dtype="string"
)

lead_df.loc[
    lead_df["Usage_Status"] == "used",
    "Reservation_Outcome"
] = "Completed"

lead_df.loc[
    lead_df["Usage_Status"].isin(["cancelled", "no-show"]),
    "Reservation_Outcome"
] = "Not Completed"

valid_lead = lead_df[
    lead_df["Lead_Time_Category"].notna()
    & (lead_df["Lead_Time_Category"] != "Unknown")
    & lead_df["Reservation_Outcome"].notna()
].copy()

lead_order = [
    "Last Minute", "Short Lead", "Medium Lead", "Early Booking"
]

lead_counts = pd.crosstab(
    valid_lead["Lead_Time_Category"],
    valid_lead["Reservation_Outcome"]
).reindex(lead_order, fill_value=0)

lead_counts = lead_counts.reindex(
    columns=["Completed", "Not Completed"], fill_value=0
)

print("\n3. RESERVATION OUTCOME BY LEAD-TIME CATEGORY")
print(lead_counts.to_string())

# Calculate outcome percentages within each lead-time group.
lead_percentages = lead_counts.div(
    lead_counts.sum(axis=1).replace(0, np.nan), axis=0
) * 100

print("\nOUTCOME PERCENTAGES WITHIN EACH LEAD-TIME CATEGORY")
print(lead_percentages.round(2).to_string())

if not lead_counts.empty:
    lead_percentages.plot(
        kind="bar",
        figsize=(10, 6)
    )
    plt.title("Reservation Outcome by Lead-Time Category")
    plt.xlabel("Lead-Time Category")
    plt.ylabel("Percentage of Reservations (%)")
    plt.xticks(rotation=30, ha="right")
    plt.ylim(0, 100)
    plt.legend(title="Reservation Outcome")
    plt.tight_layout()
    plt.show()

    completed_rates = lead_percentages["Completed"].dropna()
    if not completed_rates.empty:
        best_lead = completed_rates.idxmax()
        best_rate = completed_rates.max()
        worst_lead = completed_rates.idxmin()
        worst_rate = completed_rates.min()

        print(
            f"\nFinding: {best_lead} has the highest completion "
            f"percentage at {best_rate:.2f}%."
        )
        print(
            f"{worst_lead} has the lowest completion percentage "
            f"at {worst_rate:.2f}%."
        )
else:
    print("No valid lead-time outcome records available.")

print("\nEDA COMPLETE.")
```

```python

# ============================================
# PART IV: DATA MINING AND PREDICTIVE MODELING
# TASK 7 AND TASK 8
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# --------------------------------------------
# 1. LOAD DATASET
# --------------------------------------------

if Path("facility_reservations_cleaned.csv").exists():
    df = pd.read_csv("facility_reservations_cleaned.csv")
elif Path("facility_reservations.csv").exists():
    df = pd.read_csv("facility_reservations.csv")
elif Path("facility_reservations (1).xlsx").exists():
    df = pd.read_excel("facility_reservations (1).xlsx")
else:
    raise FileNotFoundError("Upload the reservation dataset first.")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

# Standardize column names for matching
df.columns = df.columns.str.strip()

# --------------------------------------------
# 2. CREATE THE TARGET VARIABLE
# --------------------------------------------

if "Usage_Status" not in df.columns:
    raise ValueError(
        "Usage_Status is required to construct Reservation_Outcome."
    )

status = (
    df["Usage_Status"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace({
        "canceled": "cancelled",
        "no show": "no-show",
        "noshow": "no-show"
    })
)

df["Reservation_Outcome"] = pd.Series(
    pd.NA, index=df.index, dtype="string"
)

df.loc[status == "used", "Reservation_Outcome"] = "Completed"

df.loc[
    status.isin(["cancelled", "no-show"]),
    "Reservation_Outcome"
] = "Not Completed"

# Exclude records without a known target
df = df.dropna(subset=["Reservation_Outcome"]).copy()

print("\nTarget class distribution:")
print(df["Reservation_Outcome"].value_counts())

if df["Reservation_Outcome"].nunique() < 2:
    raise ValueError(
        "The dataset has only one target class. "
        "A binary classifier cannot be evaluated meaningfully."
    )

# --------------------------------------------
# 3. SELECT PREDICTORS AVAILABLE AT BOOKING TIME
# --------------------------------------------

# Match the actual available columns in your dataset:
candidate_predictors = [
    "Facility_Type",
    "User_Category",
    "Participants",
    "Duration_Hours",
    "Lead_Days",
    "Weekday",
    "Start_Hour",
    "Facility_Capacity"
]

excluded_columns = {
    "Reservation_Outcome",
    "Usage_Status",
    "Actual_Attendance",
    "Cancellation_Days_Before",
    "Attendance_Ratio",
    "Reservation_ID"
}

features = [
    col for col in candidate_predictors
    if col in df.columns and col not in excluded_columns
]

# Remove repeated names
features = list(dict.fromkeys(features))

if len(features) < 5:
    raise ValueError(
        f"Only {len(features)} eligible predictors were found: {features}\n"
        "The examination requires at least five predictors available "
        "at reservation submission time. Inspect df.columns and add "
        "actual eligible booking-time fields to candidate_predictors. "
        "Do not use post-outcome information to reach five."
    )

X = df[features].copy()
y = df["Reservation_Outcome"].copy()

# Drop columns with no observed values
empty_columns = X.columns[X.isna().all()].tolist()

if empty_columns:
    print("\nDropping entirely empty predictors:", empty_columns)
    X = X.drop(columns=empty_columns)

features = X.columns.tolist()

if len(features) < 5:
    raise ValueError(
        "Fewer than five usable predictors remain after removing "
        "entirely empty columns. Check the actual dataset."
    )

print("\nSelected predictors:")
for feature in features:
    print("-", feature)

print("\nTarget: Reservation_Outcome")
print("Excluded post-outcome fields to prevent data leakage.")

# --------------------------------------------
# 4. IDENTIFY NUMERICAL AND CATEGORICAL FEATURES
# --------------------------------------------

numeric_features = X.select_dtypes(
    include=["number", "bool"]
).columns.tolist()

categorical_features = [
    col for col in X.columns if col not in numeric_features
]

# Convert non-numeric values in numerical columns safely
for col in numeric_features:
    X[col] = pd.to_numeric(X[col], errors="coerce")

print("\nNumerical predictors:", numeric_features)
print("Categorical predictors:", categorical_features)

# --------------------------------------------
# 5. PREPROCESSING
# --------------------------------------------

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

transformers = []

if numeric_features:
    transformers.append(
        ("numeric", numeric_pipeline, numeric_features)
    )

if categorical_features:
    transformers.append(
        ("categorical", categorical_pipeline, categorical_features)
    )

preprocessor = ColumnTransformer(transformers=transformers)

# --------------------------------------------
# 6. SPLIT DATA: 80% TRAINING, 20% TESTING
# --------------------------------------------

class_counts = y.value_counts()

# Stratification preserves class proportions when feasible.
can_stratify = (
    y.value_counts().min() >= 2
    and len(y) * 0.2 >= y.nunique()
)

stratify_arg = y if can_stratify else None

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=stratify_arg
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
print("Random state: 42")
print("Stratified split:", can_stratify)

# --------------------------------------------
# 7. BUILD AND TRAIN THE DECISION TREE
# --------------------------------------------

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(
        max_depth=5,
        min_samples_leaf=3,
        random_state=42,
        class_weight="balanced"
    ))
])

model.fit(X_train, y_train)

# --------------------------------------------
# 8. MAKE PREDICTIONS
# --------------------------------------------

y_pred = model.predict(X_test)

labels = ["Completed", "Not Completed"]

print("\nACTUAL TEST CLASS DISTRIBUTION")
print(y_test.value_counts())

print("\nPREDICTED TEST CLASS DISTRIBUTION")
print(pd.Series(y_pred).value_counts())

# --------------------------------------------
# 9. CONFUSION MATRIX
# --------------------------------------------

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=labels
)

print("\nCONFUSION MATRIX")
print(pd.DataFrame(
    cm,
    index=["Actual Completed", "Actual Not Completed"],
    columns=["Predicted Completed", "Predicted Not Completed"]
))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)
disp.plot()
plt.title("Decision Tree Confusion Matrix")
plt.tight_layout()
plt.show()

# --------------------------------------------
# 10. EVALUATION METRICS
# --------------------------------------------

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    pos_label="Not Completed",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    pos_label="Not Completed",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    pos_label="Not Completed",
    zero_division=0
)

print("\nMODEL EVALUATION")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_test,
    y_pred,
    labels=labels,
    zero_division=0
))

# --------------------------------------------
# 11. TOP THREE MOST INFLUENTIAL PREDICTORS
# --------------------------------------------

preprocessor_fitted = model.named_steps["preprocessor"]
tree = model.named_steps["classifier"]

encoded_names = preprocessor_fitted.get_feature_names_out()
importances = tree.feature_importances_

importance_df = pd.DataFrame({
    "Encoded_Feature": encoded_names,
    "Importance": importances
}).sort_values("Importance", ascending=False)

print("\nTOP 10 ENCODED FEATURES")
print(importance_df.head(10).to_string(index=False))

def original_feature(encoded_name):
    for col in sorted(features, key=len, reverse=True):
        if encoded_name.endswith("__" + col):
            return col
        if encoded_name.startswith("numeric__" + col):
            return col
        if encoded_name.startswith("categorical__" + col + "_"):
            return col
    return encoded_name

importance_df["Original_Predictor"] = (
    importance_df["Encoded_Feature"].apply(original_feature)
)

predictor_importance = (
    importance_df.groupby("Original_Predictor")["Importance"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTOP THREE MOST INFLUENTIAL PREDICTORS")
print(predictor_importance.head(3).to_string())

predictor_importance.head(10).sort_values().plot(kind="barh")
plt.title("Decision Tree Feature Importance")
plt.xlabel("Aggregated Feature Importance")
plt.ylabel("Original Predictor")
plt.tight_layout()
plt.show()

# --------------------------------------------
# 12. VISUALIZE THE TRAINED DECISION TREE
# --------------------------------------------

plt.figure(figsize=(20, 10))
plot_tree(
    tree,
    feature_names=encoded_names,
    class_names=tree.classes_,
    filled=True,
    rounded=True,
    fontsize=8
)
plt.title("Trained Decision Tree for Reservation Outcome")
plt.tight_layout()
plt.show()

# --------------------------------------------
# 13. SAVE TEST PREDICTIONS
# --------------------------------------------

results = X_test.copy()
results["Actual_Outcome"] = y_test
results["Predicted_Outcome"] = y_pred

results.to_csv("reservation_model_predictions.csv", index=False)
importance_df.to_csv("reservation_feature_importance.csv", index=False)

print("\nSaved files:")
print("- reservation_model_predictions.csv")
print("- reservation_feature_importance.csv")
```
